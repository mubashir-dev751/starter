```python
I am choosing Lane 2: Refresh / Content Opportunity Scoring.
A model is never the goal; a better decision is. The business problem here is a lack of editorial capacity, not a lack of data. The system currently uses static baseline rules which flag too many pages, creating an unmanageable queue. By scoring and ranking these opportunities, we transition from a noisy "rule-based alert" system to a prioritized decision-support tool.
```

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
The Decision: Which declining pages should human editors review first to maximize potential traffic recovery?

The Action: A human reviewer inspects the top-ranked pages to trigger an update, rewrite, or metadata optimization.

Cost of a Wrong Call: Wasted editorial hours spent rewriting a page that dropped due to uncontrollable factors (like seasonality or market consolidation), while a high-value, genuinely decaying page is ignored.

Why ML helps: Static rules flag every drop equally. ML earns its place because the pattern of true decay is tangled across multiple signals (age, impression volume, position tiers, and engagement rates) that are too messy to write by hand.

The Frame: For human editors, deciding which declining pages to review first, we will build a ranked queue from observable search and content signals, scoring decay risk measured by Precision@50. A wrong call costs wasted editorial hours. A plain rule isn't enough because it flags too many false positives. We will claim only decision-support results.
```
